4. Create a class named Rectangle with attributes: 
o length  
o width  
           Create a method to calculate and print the area of the rectangle. 

```python
class Rectangle:
    def __init__(self,length,width):
        self.length = length
        self.width = width
    def area(self):
        area = self.length*self.width
        print(area) 
r = Rectangle(10,2)
r.area()
    
```
