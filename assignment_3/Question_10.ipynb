10. Create a class named LibraryBook with attributes:  
● book_name  
● author  
● price  
            Create a method display_book_info() that prints all book details. Create    
at least three book objects and display their information. 

```python
class LibraryBook:
    def __init__(self,book_name,author,price):
        self.book_name = book_name
        self.author = author
        self.price = price
    def display_book_info(self):
        print("Book Name:", self.book_name)
        print("Author:", self.author)
        print("Price:", self.price)
b1 = LibraryBook("Atomic Habits", "James Clear",200)
b2 = LibraryBook("Alchemist", "Paulo Coelho", 500)
b3 = LibraryBook("Deception Point", "Dan Brown", 300)
b1.display_book_info()
b2.display_book_info()
b3.display_book_info()
```
