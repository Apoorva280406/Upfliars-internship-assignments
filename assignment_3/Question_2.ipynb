2. Create a class named Car with attributes: 
o brand  
o model  
o price  
Create two objects and display their details. 

```python
class Car:
 def __init__(self,brand,model,price):
    self.brand = brand
    self.model = model
    self.price = price
c1 = Car("Toyota","Fortuner",3500000)
c2 = Car("Hyundai","Creta",1800000)
print(c1.brand,c1.model,c1.price)
print(c2.brand,c2.model,c2.price)
```
