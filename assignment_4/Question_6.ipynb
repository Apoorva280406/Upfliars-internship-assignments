6. Create at least two objects and display their grades.

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks
    def calculate_grade(self):
        if self.marks >= 90:
            grade = "A"
        elif self.marks >= 75:
            grade = "B"
        elif self.marks >= 60:
            grade = "C"
        else:
            grade = "D"
        return grade
s1 = Student("Apoorva", 78)
s2 = Student("Aanya", 87)
print("Name:", s1.name)
print("Marks:", s1.marks)
print("Grade:", s1.calculate_grade())
print()
print("Name:", s2.name)
print("Marks:", s2.marks)
print("Grade:", s2.calculate_grade())
```
