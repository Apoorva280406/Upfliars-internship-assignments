4. Create a function that calculates simple interest. 
 Use default arguments for rate and time. Demonstrate calling the function 
 using both positional and keyword arguments.

```python
def simple_interest(principal, rate=5, time=2):
    si = (principal * rate * time) / 100
    return si
interest1 = simple_interest(10000, 8, 3)
print("Simple Interest:", interest1)
interest2 = simple_interest(principal=10000, rate=10, time=2)
print("Simple Interest:", interest2)
interest3 = simple_interest(10000)
print("Simple Interest:", interest3)
```
