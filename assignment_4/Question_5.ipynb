5.Create a Student class with attributes: name, marks. Add a method to 
 calculate grade based on marks.

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks
    def calculate_grade(self):
        if self.marks >= 90:
            grade = "A"
        elif self.marks >= 75:
            grade = "B"
        elif self.marks >= 60:
            grade = "C"
        else:
            grade = "D"
        return grade
s1 = Student("Apoorva", 90)
print("Name:", s1.name)
print("Marks:", s1.marks)
print("Grade:", s1.calculate_grade())
```
