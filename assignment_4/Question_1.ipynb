1. Write a program that takes a list of numbers and prints all prime 
 numbers in the list using loops. Do not use any built-in prime-checking 
 functions. 

```python
numbers = [10, 7, 13, 20, 17, 8, 19]
for num in numbers:
    if num > 1:
        is_prime = True
        for i in range(2, num):
            if num % i == 0:
                is_prime = False
                break

        if is_prime:
            print(num)
```
