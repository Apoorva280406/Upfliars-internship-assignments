5. Create a DataFrame of at least 10 students with columns: Name, Age, 
Marks, City. Perform the following:  
• Display the first 5 rows using head() 
• Show basic statistics using describe() 
• Filter students who scored more than 75 marks 
• Sort by Marks in descending order 


```python
import numpy as np
import pandas as pd
```

```python
data = {
    "Name": ["Apoorva", "Priyam", "Divya", "Arpit", "Karan",
             "Neha", "Swati", "Pooja", "Unnati", "Riya"],
    "Age": [20, 21, 19, 22, 20, 21, 23, 20, 22, 19],
    "Marks": [85, 72, 90, 68, 77, 81, 95, 74, 88, 79],
    "City": ["Bikaner", "Pilani", "Jaipur", "Kota", "Delhi",
             "Churu", "Bangalore", "Kolkata", "Hyderabad", "Jaipur"]
}
df = pd.DataFrame(data)
print("First 5 rows: ",df.head(5))
print("Basic stat: ",df.describe())
print("Students who scorsd more than 75 marks: \n",df[df["Marks"]>75])
print("Marks in descending order: \n",df.sort_values("Marks", ascending = False))
```
