12) Create a subplot layout of 2 rows × 2 columns using Matplotlib 
containing:  
• Line plot 
• Scatter plot 
• Bar chart 
• Histogram 
All using the same dataset. Add a main title using suptitle()

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    "Student": ["Apoorva", "Priyam", "Divya", "Arpit", "Karan"],
    "Marks": [85, 92, 78, 88, 95],
    "Age": [20, 21, 19, 22, 20]
}

df = pd.DataFrame(data)
fig, axes = plt.subplots(2, 2, figsize=(10, 8))
axes[0, 0].plot(df["Student"], df["Marks"], marker="o")
axes[0, 0].set_title("Line Plot")
axes[0, 0].set_xlabel("Student")
axes[0, 0].set_ylabel("Marks")
axes[0, 1].scatter(df["Age"], df["Marks"])
axes[0, 1].set_title("Scatter Plot")
axes[0, 1].set_xlabel("Age")
axes[0, 1].set_ylabel("Marks")
axes[1, 0].bar(df["Student"], df["Marks"])
axes[1, 0].set_title("Bar Chart")
axes[1, 0].set_xlabel("Student")
axes[1, 0].set_ylabel("Marks")
axes[1, 1].hist(df["Marks"], bins=5)
axes[1, 1].set_title("Histogram")
axes[1, 1].set_xlabel("Marks")
axes[1, 1].set_ylabel("Frequency")
plt.suptitle("Student Dataset Visualization", fontsize=16)
plt.tight_layout()
plt.show()

```
