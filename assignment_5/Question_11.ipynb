11)Using Seaborn, create a dataset of at least 50 records with columns 
Age, Salary, Department, Experience. Plot:  
• A scatter plot of Age vs Salary with hue as Department 
• A heatmap showing correlation between numeric columns 
• A bar plot showing average Salary per Department 

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
np.random.seed(42)
df = pd.DataFrame({
    "Age": np.random.randint(22, 50, 50),
    "Salary": np.random.randint(30000, 100000, 50),
    "Department": np.random.choice(
        ["HR", "IT", "Finance", "Marketing"], 50
    ),
    "Experience": np.random.randint(1, 20, 50)
})
print(df.head())
sns.scatterplot(
    data=df,
    x="Age",
    y="Salary",
    hue="Department"
)
plt.title("Age vs Salary by Department")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.show()
correlation_matrix = df[["Age", "Salary", "Experience"]].corr()

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm"
)
plt.title("Correlation Heatmap")
plt.show()
sns.barplot(
    data=df,
    x="Department",
    y="Salary",
    estimator=np.mean
)

plt.title("Average Salary per Department")
plt.xlabel("Department")
plt.ylabel("Average Salary")
plt.show()
```
