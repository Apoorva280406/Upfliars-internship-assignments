9. Using Matplotlib, create the following plots from a student marks 
dataset (create the dataset yourself):  
• A line plot showing marks trend across 5 subjects 
• A bar chart comparing marks of 5 students 
• Add proper title, xlabel, ylabel, legend, and grid to both plots 

```python
import matplotlib.pyplot as plt
```

```python
subjects = ["Math", "Science", "English", "SST", "Hindi"]
marks = [85, 90, 78, 95, 88]
students = ["Apoorva","Priyam","Divya","Arpit","Riya"]
plt.plot(subjects, marks)
plt.title("Marks Trend Across 5 Subjects")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.legend()
plt.grid(True)
plt.show()
plt.bar(students,marks)
plt.title("Marks Trend Across 5 Students")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.legend()
plt.grid(True)
plt.show()



```
