8. Merge two DataFrames:  
• DataFrame 1: StudentID, Name, Department 
• DataFrame 2: StudentID, Marks, Grade 
• Merge on StudentID and display students who scored Grade 'A' 

```python
import numpy as np
import pandas as pd
```

```python
df1 = pd.DataFrame({
    "StudentID": [101, 102, 103, 104, 105],
    "Name": ["Apoorva", "Priyam", "Divya", "Swati", "Arpit"],
    "Department": ["AI", "IT", "CSE", "ECE", "DS"]})
df2 = pd.DataFrame({
    "StudentID": [101, 102, 103, 104, 105],
    "Marks": [92, 78, 88, 95, 70],
    "Grade": ["A", "B", "A", "A", "C"]
})
merge = pd.merge(df1,df2,on="StudentID")
print("Merged dataframe: \n",merge)
a_grade = merge[merge["Grade"]=="A"]
print("Grade A students: \n",a_grade)
```
