4. Using np.linspace() and np.zeros() and np.ones(), create three different 
arrays and stack them vertically using np.vstack(). Display the final 
stacked array.

```python
import numpy as np
```

```python
a = np.linspace(1,4,3)
b = np.zeros(3)
c = np.ones(3)
print("Array from np.linspace(): ",a)
print("Array from np.zeros(): ",b)
print("Array from np.ones(): ",c)
final_array=np.vstack((a,b,c))
print("Final array: ",final_array)
```
