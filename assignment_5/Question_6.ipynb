6. Create a DataFrame with some missing values intentionally placed in the 
Marks and City columns. Then:  
• Identify missing values using isnull() 
• Fill missing marks with the column mean using fillna() 
• Drop rows where City is missing using dropna()

```python
import numpy as np
import pandas as pd
```

```python
data = {
    "Name": ["Apoorva", "Priyam", "Divya", "Arpit", "Karan"],
    "Marks": [85, np.nan, 90, 78, np.nan],
    "City": ["Delhi", "Mumbai", np.nan, "Pune", np.nan]
}

df = pd.DataFrame(data)
df.isnull()
df.fillna({'Marks': df['Marks'].mean()},inplace=True)
print("data after filling missing marks with mean:\n ",df)
df.dropna(subset = ["City"],inplace=True)
print("Data after dropping rows with missing City:\n",df)
```
